```python
source_file = f"{landing_folder_path}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```


## Step 1 - Read the JSON file using the dataframe API

```python
# Define the schema
constructors_schema = "constructorId STRING, name STRING, nationality STRING, url STRING"
```

```python
constructors_df = (
    spark.read
        .format('json')
        .schema(constructors_schema)
        .option('mode', 'FAILFAST')
        .load(source_file)
)
```

```python
display(constructors_df)
```

## Add Metadata Columns
- Source File
- Ingestion Timestamp


```python
constructors_final_df = add_injestion_metadata(constructors_df)
```

## Write to bronze delta table

```python
(
    constructors_final_df
        .write
        .format('delta')
        .mode('overwrite')
        .saveAsTable(table_name)
)
```
