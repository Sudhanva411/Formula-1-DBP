# Project Enviroment setup for Formula 1

1. Create external location db-ext-dl1-formula1
2. Create Catalog formula1
3. Create Schemas landing, bronze, silver and gold
4. Create Volume Files in landing schema


## Access Cloud Storage

```sql
%fs ls 'abfss://formula1@dbcourseextdl01.dfs.core.windows.net/landing/'
```

## Create External Location db-ext-dl1-formula1

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS db_ext_dl1_formula1
URL 'abfss://formula1@dbcourseextdl01.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `db-course-sc`)
COMMENT 'External location for formula 1 container'
```

## Create Catalog formula1

```sql
CREATE CATALOG IF NOT EXISTS formula1
     MANAGED LOCATION 'abfss://formula1@dbcourseextdl01.dfs.core.windows.net/' 
     COMMENT 'This is the main catalog for the formula1 project' ;
```

```sql
SHOW CATALOGS;
```

## Create Schemas landing, bronze, silver and gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@dbcourseextdl01.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@dbcourseextdl01.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@dbcourseextdl01.dfs.core.windows.net/gold';
```

```sql
USE CATALOG formula1;
SELECT current_catalog();
```

```sql
SHOW SCHEMAS;
```

## Create Volume Files in landing schema

```sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@dbcourseextdl01.dfs.core.windows.net/landing';
```

```sql
%fs ls /Volumes/formula1/landing/files
```
